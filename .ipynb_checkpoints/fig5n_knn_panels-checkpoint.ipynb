```python
# Function to load experiments with specified thresholds and z-normalization
def exp_loader(exp_name, normalize=True, normalization_condition='B50'):
    exp_parameters = {
        'Z2': ('250930_Perturbation6iExpZ2', 'expZ2_data_scaled_bgsub_6_cols_251002.csv',
               {'SOX17_1': 'SOX17', 'TBX6_2': 'TBX6', 'SOX2_2': 'SOX2', 'DAPI_1': 'DAPI'},
               ['B50', 'B50I5', 'B50SB10', 'B50M0p5'],
               {'ISL1': 200, 'TFAP2C': 350, 'SOX17': 200, 'NANOG': 800, 'TBX6': 300, 'TBXT': 175, 'SOX2': 150},
               'expZ2_junk_labels_nans_removed.csv'),
        'Z1B': ('250911_Perturbation6iExpZ1B', '250911_ExpZ1B_scaled.csv',
               {'SOX17_1': 'SOX17', 'DAPI_1': 'DAPI'},
               ['B50', 'B50I5', 'B50SB10', 'B50_MEKi'],
               {'ISL1': 200, 'TFAP2C': 350, 'SOX17': 200, 'NANOG': 800, 'TBX6': 300, 'TBXT': 175, 'SOX2': 150},
               None),
        'X7': ('250909_Perturbation6iExpX7', 'expX7_data_scaled_bgsub_4_cols_250909.csv',
               {'SOX17_1': 'SOX17', 'DAPI_1': 'DAPI'},
               ['B50', 'B50T30'],
               {'ISL1': 150, 'TFAP2C': 400, 'SOX17': 200, 'NANOG': 600, 'TBX6': 400, 'TBXT': 175, 'SOX2': 250},
               None),
        '20': ('250509_Exp20', 'exp20_data_bgsub_scaled_250509.csv',
               {'SOX17_1': 'SOX17', 'DAPI_1': 'DAPI'},
               ['B50', 'B200', 'B10'],
               {'ISL1': 500, 'GATA3': 700, 'HAND1': 500, 'CDX2': 600, 'TFAP2C': 600, 'SOX17': 500, 'PRDM1': 650, 
                'SNAI1': 300, 'MIXL1': 400, 'EOMES': 200, 'TBXT': 285, 'TBX6': 375,  'OCT4': 500, 'NANOG': 1600, 'SOX2': 400, 'OTX2': 400},
               None),
        '28': ('250722_Exp28', 'exp28_data_bgsub_scaled_251211.csv',
               {'SOX17_1': 'SOX17', 'DAPI_1': 'DAPI'},
               ['B50', 'B200', 'B10'],
               {'ISL1': 200, 'TFAP2C': 600, 'SOX17': 250, 'NANOG': 1000, 'TBX6': 1000, 'TBXT': 305, 'SOX2': 200},
               None)
    }
    
    base_dir = 'Z:/Shared-03/BF/Figures_4i/signaldecoding'
    signal_names = ['pSMAD1', 'ncYAP', 'LEF1', 'ncSMAD23', 'cytopERK']
    feature_names = ['nucArea', 'nucVolume', 'nucMajorAxis', 'nucMinorAxis', 
                    'nucCircularity', 'CircleEdgeDist', 'RadialDist']
    
    exp_dir, csv_file, rename_dict, conditions, thresh, junk_csv = exp_parameters[exp_name]
    data_dir = os.path.join(base_dir, exp_dir)
    
    # Load and clean data
    data = pd.read_csv(os.path.join(data_dir, csv_file))
    data.rename(columns=rename_dict, inplace=True)
    
    if junk_csv:
        junk = pd.read_csv(os.path.join(data_dir, junk_csv)) == 2
        data = data[~junk.iloc[:, 0].values]
    
    gene_names = list(thresh.keys())
    valid_mask = ~(data[signal_names + gene_names].isna() | (data[signal_names + gene_names] < 0)).any(axis=1)
    data = data[valid_mask].reset_index(drop=True)


    # --- Z-normalization ---
    if normalize:
        # Use only the specified condition for statistics (usually B50)
        norm_mask = (data['condition'] == normalization_condition)
        norm_data = data.loc[norm_mask]
        norm_cols = signal_names + gene_names
        
        norm_mean = norm_data[norm_cols].mean()
        norm_std = norm_data[norm_cols].std()
        
        # Normalize all signals and genes
        data_z = data.copy()
        data_z[norm_cols] = (data[norm_cols] - norm_mean) / norm_std

    else:
        data_z = data.copy()
    
    # Create metadata
    meta = IIF.Metadata()
    meta.xres = meta.yres = 0.325
    meta.channels = signal_names + gene_names
    meta.conditions = conditions
    
    return base_dir, data_dir, data, data_z, thresh, meta, gene_names, signal_names, feature_names


# Create colony and exp instances for a given dataset
def create_experiment(data, meta, feature_names):
    colonies = {}
    for colID in data['Colony'].unique():
        col = IIF.Colony(data[data['Colony'] == colID], colID, meta, 
                        features=feature_names, nominalRadius=350)
        col.calcRadialProfiles(cellsPerBin=50, overlap=0, dr=8)
        col.calcPosError(sigma=1)
        colonies[colID] = col
    
    exp = IIF.MPexperiment(colonies, meta)
    exp.calcRadialProfiles(cellsPerBin=50, overlap=0)
    return colonies, exp


# Save colony and experiments as a ExperimentResults object
class ExperimentResults:
    def __init__(self, exp_id):
        # Load data
        base_dir, data_dir, data, data_z, thresh, meta, gene_names, signal_names, feature_names = exp_loader(exp_id)
        
        self.meta = meta
        self.signal_names = signal_names
        self.gene_names = gene_names
        self.feature_names = feature_names
        self.thresh = thresh
        self.data = data
        self.data_z = data_z
        self.base_dir = base_dir
        self.data_dir = data_dir
        
        # Create experiments
        self.col, self.exp = create_experiment(self.data, meta, feature_names)
        self.col_z, self.exp_z = create_experiment(self.data_z, meta, feature_names)
```

```python
# Apply various smoothing methods to signal data
def apply_smoothing(signals, k_val, mode='gaussian'):
    
    distances, indices = NearestNeighbors(n_neighbors= k_val, n_jobs=-1).fit(
        StandardScaler().fit_transform(signals)).kneighbors()
    
    if mode == 'simple':
        neighbor_signals = signals[indices[:, :k_val]]
        smoothed = neighbor_signals.mean(axis=1)
        
    elif mode == 'gaussian':
        neighbor_signals = signals[indices[:, :k_val]]
        sigma = np.maximum(distances[:, k_val-1:k_val], 1e-10)
        weights = np.exp(-0.5 * (distances[:, :k_val] / sigma) ** 2)
        smoothed = np.einsum('ij,ijk->ik', 
                                weights / weights.sum(axis=1, keepdims=True), 
                                neighbor_signals)
    
    return smoothed
```

```python
# Create perturbed dataset using KNN from low-perturbation cells
def create_perturbed_data(data, exp, signal_names, perturbing_sigs, perturb_mode='inh', do_naive = True, use_profile = True,
                         percentile=1, k_nn=5, distance_signals=None, preserve_signals=None):

    # Z-normalize
    data_z = data.copy()
    data_z[signal_names] = (data[signal_names] - data[signal_names].mean()) / data[signal_names].std()

    # Define subset based on perturbation
    if perturb_mode == 'inh':
        if use_profile:
            threshold = np.min(exp.radialProfiles['B50'][perturbing_sigs])
        else:
            threshold = np.percentile(data[perturbing_sigs], percentile)
            print(threshold)
        subset_mask = (data[perturbing_sigs].values <= threshold).all(axis=1)
    elif perturb_mode == 'OA':
        if use_profile:
            threshold = np.max(exp.radialProfiles['B50'][perturbing_sigs])
        else:
            threshold = np.percentile(data[perturbing_sigs], 100 - percentile)
        subset_mask = (data[perturbing_sigs].values >= threshold).all(axis=1)
        
    else:
        raise ValueError(f"Unknown perturb_mode: {perturb_mode}")

    if do_naive:
        data_perturbed = data.copy()
        data_perturbed[perturbing_sigs] = threshold

        data_perturbed_z = (data_perturbed[signal_names] - data[signal_names].mean()) / data[signal_names].std()
        return data_perturbed, data_perturbed_z

    else:
        data_subset = data[subset_mask]
        data_z_subset = data_z[subset_mask]
        
        print(f"Perturbation subset: {subset_mask.sum()} / {len(data)} cells ({100*subset_mask.mean():.1f}%)")
        
        # Distance signals
        if distance_signals is None:
            distance_signals = [s for s in signal_names if s not in perturbing_sigs]
        
        # Fit KNN and find neighbors
        knn_model = NearestNeighbors(n_neighbors=k_nn, metric='euclidean')
        knn_model.fit(data_z_subset[distance_signals])
        distances, indices = knn_model.kneighbors(data_z[distance_signals])
        
        # Gaussian weighted averaging
        subset_signals = data_subset[signal_names].values
        sigma = np.maximum(distances[:, -1:], 1e-10)
        weights = np.exp(-0.5 * (distances / sigma) ** 2)
        mean_signals = np.einsum('ij,ijk->ik', 
                                weights / weights.sum(axis=1, keepdims=True), 
                                subset_signals[indices])

        # Gaussian weighted averaging (z_norm)
        subset_signals_z = data_z_subset[signal_names].values
        sigma = np.maximum(distances[:, -1:], 1e-10)
        weights = np.exp(-0.5 * (distances / sigma) ** 2)
        mean_signals_z = np.einsum('ij,ijk->ik', 
                                weights / weights.sum(axis=1, keepdims=True), 
                                subset_signals_z[indices])
        
        # Create output
        data_perturbed = data.copy()
        data_perturbed[signal_names] = mean_signals

        # Create z-normalized output
        data_perturbed_z = data_z.copy()
        data_perturbed_z[signal_names] = mean_signals_z
        
        # Optional: Preserve specific signals in control
        if preserve_signals:
            for sig in preserve_signals:
                if sig in signal_names:
                    data_perturbed[sig] = data[sig]
                    data_perturbed_z[sig] = data_z[sig]
        
        return data_perturbed, data_perturbed_z
```

```python
def run_vib_predictions(data_dir, data_train, data_test, signal_names, gene_names, 
                       hyperparam, N_run = 20, output_prefix='vib_pred'):
    
    feat_names = signal_names
    vib = {}
    predictions = {}

    feat_train = data_train[feat_names]
    print(np.shape(feat_train))
    tar_train = data_train[gene_names]

    feat_test = data_test[feat_names]
    print(np.shape(feat_test))
    
    for run in range(N_run):
        #print(f"  VIB Run {run+1}/{N_run}")
        start = time.time()

        # set seeds for reproducibility
        torch.manual_seed(run)
        np.random.seed(run)
        if torch.backends.mps.is_available():
            torch.mps.manual_seed(run)
    
        vib[run] = IIF.VIB(feat_train, tar_train, hyperparam)
    
        if not os.path.exists(data_dir + '/model_' + str(run) + '_' + str(len(signal_names)) + 'D' + '.pth'):
            print('run VIB')
            _ = vib[run].train(verbose=False)
            torch.save(vib[run].model, data_dir + '/model_' + str(run) + '_' + str(len(signal_names)) + 'D' + '.pth')
        else:
            print('load VIB')
            vib[run].model = torch.load(data_dir + '/model_' + str(run) + '_' + str(len(signal_names)) + 'D' + '.pth', map_location=torch.device('cpu'), weights_only=False)
        
        # Predict
        tar_predict = vib[run].predict(feat_test)
        
        pred_df = data_test.copy()
        pred_df.loc[tar_predict.index, tar_predict.columns] = tar_predict
        predictions[run] = pred_df
        
        # Save individual run
        filename = (f"{output_prefix}_{len(feat_names)}D_"
                   f"Nhid{hyperparam['HIDDEN_DIM']}_"
                   f"Nlay{hyperparam['N_LAYERS']}_"
                   f"Ldim{hyperparam['LATENT_DIM']}_run{run}.csv")
        #pred_df.to_csv(filename, index=False)
        print(f"    Time: {time.time() - start:.1f}s")
    
    # Average predictions
    avg_pred = predictions[0].copy()
    avg_genes = pd.concat([df[gene_names] for df in predictions.values()]).groupby(level=0).mean()
    avg_pred[gene_names] = avg_genes
    predictions['avg'] = avg_pred
    
    return vib, avg_pred, predictions
```

```python
def fate_histogram(data_cond, exp_result):
    n_bins = 20  # adjust this number as needed
    thresh_boundary = 0.5 # boundaries are half maximum of fate
    colors = list(IIF.colmap_fates.values())
    fate_names_ordered = list(IIF.colmap_fates.keys())
    N_fates = len(fate_names_ordered)
        
    fates, fate_names = IIF.return_fates(data_cond, thresh=exp_result.thresh)
    
    # bin by radial distance
    binning = pd.qcut(data_cond['CircleEdgeDist'], q=n_bins, labels=False, duplicates='drop').to_frame(name='r_bin')
    binning['fate'] = fates

    # get bin centers in microns
    bin_edges = pd.qcut(data_cond['CircleEdgeDist'], q=n_bins, duplicates='drop', retbins=True)[1]
    bin_centers = (bin_edges[1:] + bin_edges[:-1]) / 2

    # fate counts per bin
    fate_counts = binning.groupby(['r_bin', 'fate']).size().unstack(fill_value=0)
    fate_counts = fate_counts.reindex(columns=fate_names_ordered, fill_value=0)

    # normalize per fate (each fate peaks at 1) for boundary detection
    fate_counts_byfate = fate_counts.div(fate_counts.sum(axis=0), axis=1)
    fate_counts_byfate_norm = fate_counts_byfate.div(fate_counts_byfate.max(axis=0).replace(0, 1), axis=1)

    # normalize per bin for bar plot
    fate_counts_norm = fate_counts.div(fate_counts.sum(axis=1), axis=0)
    
    # create the stacked bar graph
    # distribution bar plot
    fig, ax = plt.subplots(1, 1, figsize=(4, 3))
    fate_counts_norm.plot(kind='bar', color=colors, stacked=True, ax=ax, width=1, legend=False)
    #ax.set_title(cond)
    ax.set_xticks([0, n_bins-1], labels=['edge', 'center'], rotation=0)
    ax.set_yticks([])
    ax.set_ylim(0, 1)
    ax.set_xlabel(None)
    ax.set_xlim(-0.5, n_bins - 0.5)
    ax.set_box_aspect(0.5)
    sw = 2
    for spine in ax.spines.values():
        spine.set_linewidth(sw)
    

    plt.tight_layout()
```

```python
def plot_colony_pie_triplet(colony_list, exp_result_list, labels,
                             ms=7, figsize=(7, 7), start_angle=90,
                             boundary_color='black', boundary_lw=2, lim = 375, dpi=150):
    """
    Save colony scatter plots as images, crop whitespace,
    clip each to its sector, and arrange in a three-way pie plot.
    """
    import io
    from matplotlib.patches import Wedge
    
    sector_starts = [start_angle + i * 120 for i in range(3)]
    
    fig, ax = plt.subplots(1, 1, figsize=(5, 5), constrained_layout=True)
    #ax.set_xlim(-1.2, 1.2)
    #ax.set_ylim(-1.2, 1.2)
    ax.set_aspect('equal')
    ax.axis('off')
    
    for colony, exp_result, label, s_start in zip(
        colony_list, exp_result_list, labels, sector_starts
    ):

        
        fig_tmp, ax_tmp = plt.subplots(1, 1, figsize=(5, 5), constrained_layout=True)
        colony.scatter_fates(ms=ms, ax=ax_tmp, legend=False, thresh=exp_result.thresh)
        #plt.tight_layout()
        
        buf = io.BytesIO()
        fig_tmp.savefig(buf, format='png', dpi=dpi, bbox_inches='tight',
                       transparent=True)
        buf.seek(0)
        img = plt.imread(buf)
        plt.close(fig_tmp)
        
        # ====================================================================
        # Crop whitespace
        # ====================================================================
        # Find rows and columns that are not white/transparent
        if img.shape[2] == 4:
            # RGBA: use alpha channel
            mask = img[:, :, 3] > 0.01
        else:
            # RGB: use all channels (white = [1, 1, 1])
            mask = ~(np.all(img > 0.99, axis=2))
        
        rows = np.any(mask, axis=1)
        cols = np.any(mask, axis=0)
        
        rmin, rmax = np.where(rows)[0][[0, -1]]
        cmin, cmax = np.where(cols)[0][[0, -1]]
        
        # Add small padding
        pad = 5
        rmin = max(0, rmin - pad)
        rmax = min(img.shape[0], rmax + pad)
        cmin = max(0, cmin - pad)
        cmax = min(img.shape[1], cmax + pad)
        
        img_cropped = img[rmin:rmax, cmin:cmax]
        
        # ====================================================================
        # Place image at center, clipped to sector wedge
        # ====================================================================
        wedge = Wedge(center=(0, 0), r=1.0,
                     theta1=s_start, theta2=s_start + 120,
                     transform=ax.transData)
        
        im = ax.imshow(img_cropped, extent=[-1, 1, -1, 1],
                      aspect='equal', zorder=2,
                      origin='upper')
        im.set_clip_path(wedge)
        
        # # Sector label
        # mid_rad = np.radians(s_start + 60)
        # ax.text(1.1 * np.cos(mid_rad), 1.1 * np.sin(mid_rad),
        #         label, ha='center', va='center', fontsize=12, fontweight='bold')
    
    # Sector dividers
    for angle in sector_starts:
        rad = np.radians(angle)
        ax.plot([0, np.cos(rad)], [0, np.sin(rad)],
                color=boundary_color, linewidth=boundary_lw, zorder=5)
    
    return fig, ax

def encode_data(vib, data_z, signal_names):
    """Encode data into latent space and return mu"""
    with torch.no_grad():
        mu, _ = vib.model.encode(torch.FloatTensor(data_z[signal_names].values))
    return mu.numpy()
```

```python
# # Section 3: VISUALIZE LATENT SPACE
X = vib_choice.scaler_X_run.transform(exp_result_model.data.loc[mask, exp_result_model.signal_names])
X = torch.FloatTensor(X).to(device)

# 2D model
model.eval() 
with torch.no_grad():
    mu, _ = model.encode(X)
    Y = model.decode(mu).cpu().numpy()
    Y = vib_choice.scaler_Y_run.inverse_transform(Y)
    Y = pd.DataFrame(Y, columns=exp_result_model.gene_names, index=exp_result_model.data[mask].index)

L1 = mu[:, 0].cpu().numpy()
L2 = mu[:, 1].cpu().numpy()

# Standardize target
#scaler_y = StandardScaler()
L1mean = L1.mean()
L2mean = L2.mean()
L1std = L1.std()

# scale by the same factor to maintain proportions
L1_scaled = (L1 - L1mean)/L1std
L2_scaled = (L2 - L2mean)/L1std

#=============PLOT=================

fs=26
sw = 2
n=1
m=1
s = 5
alpha = 0.5

for colorfor in ['exp','pred']:

    fig, ax = plt.subplots(n, m, figsize=(5*m, 5*n))

    if colorfor == 'exp':
        fates,_ = IIF.return_fates(exp_result_model.data[mask],thresh=exp_result_model.thresh)
        colors = [IIF.colmap_fates[f] for f in fates]
        ax.scatter(L1_scaled, L2_scaled,c=colors, s=s, alpha=alpha)
    elif colorfor == 'pred':
        fates_pred,_ = IIF.return_fates(Y,thresh=exp_result_model.thresh)
        colors = [IIF.colmap_fates[f] for f in fates_pred]
        ax.scatter(L1_scaled, L2_scaled,c=colors, s=s, alpha=alpha)

    
    ax.set_xlabel('Latent Coordinate 1',fontsize=fs)
    ax.set_ylabel('Latent Coordinate 2',fontsize=fs)
    ax.set_xticks([]);
    ax.set_yticks([]);
    
    for spine in ax.spines.values():
        spine.set_linewidth(sw)

    # Make plot square by extending axis limits without changing aspect ratio
    marg = 0.5
    xlim = (np.percentile(L1_scaled, 0.1) - marg, np.percentile(L1_scaled, 99.9) + marg)
    ylim = (np.percentile(L2_scaled, 0.1) - marg, np.percentile(L2_scaled, 99.9) + marg)
    
    # xlim = (L1_scaled.min(), L1_scaled.max()) #ax[run].get_xlim()
    # ylim = (L2_scaled.min(), L2_scaled.max()) #ax[run].get_ylim() 
    
    x_range = xlim[1] - xlim[0]
    y_range = ylim[1] - ylim[0]
    
    # Find the larger range
    max_range = max(x_range, y_range)
    
    # Calculate centers
    x_center = (xlim[0] + xlim[1]) / 2
    y_center = (ylim[0] + ylim[1]) / 2
    
    L1smin = x_center - max_range/2
    L1smax = x_center + max_range/2
    L2smin = y_center - max_range/2
    L2smax = y_center + max_range/2
    
    # Set new limits centered around the data with equal range
    ax.set_xlim(L1smin, L1smax)
    ax.set_ylim(L2smin, L2smax)
    
    ax.set_aspect('equal', adjustable='box')
    plt.tight_layout(pad=0.5)
    
    #plt.savefig(dataDir+'/LS_fate_' + str(len(signals)) + 'D' + '_run' + str(run) + '_' + colorfor+'.png')
```

```python
# EXTRAPOLATE FATE MAP
l1 = np.linspace(L1smin, L1smax, 200)
l2 = np.linspace(L2smin, L2smax, 200)
L1grid_scaled, L2grid_scaled = np.meshgrid(l1, l2)


mugrid = np.stack([L1grid_scaled.ravel(), L2grid_scaled.ravel()],axis=1) # this said just L1grid before (not _scaled)
mugrid = torch.FloatTensor(mugrid).to(device)

model.eval() 
with torch.no_grad():
    Y = model.decode(mugrid).cpu().numpy()
    Y = vib_choice.scaler_Y_run.inverse_transform(Y)
    Y = pd.DataFrame(Y, columns=exp_result_model.gene_names)
    
fates,_ = IIF.return_fates(Y,thresh=exp_result_model.thresh)

#---------------------------

# data to overlay
cond = 'B50'
mask = (exp_result_20.data['condition'] == cond)

X = vib_choice.scaler_X_run.transform(exp_result_model.data.loc[mask, exp_result_model.signal_names])
X = torch.FloatTensor(X).to(device)

# 2D model
model.eval() 
with torch.no_grad():
    mu, _ = model.encode(X)

L1 = mu[:, 0].cpu().numpy()
L2 = mu[:, 1].cpu().numpy()

# Standardize target
#scaler_y = StandardScaler()
L1_scaled = (L1 - L1.mean())/L1.std()
L2_scaled = (L2 - L2.mean())/L1.std()
```

```python
#--------PLOT --------------

fs=26
sw = 2
n=1
m=1
s = 5

colors = ['white', 'black']

# Encode latent spaces
if do_naive:
    mu_perturb = encode_data(vib_choice, data_naive_z, exp_result.signal_names)
    labels = ['Exp20 B50', f'Exp20 naive {PERTURB_KWARGS['perturbing_sigs'][0]} {PERTURB_KWARGS['perturb_mode']}']
    filename = '/LS_fate_'  + str(len(exp_result.signal_names))  + f'D_{PERTURB_KWARGS['perturbing_sigs'][0]}_{PERTURB_KWARGS['perturb_mode']}_ctrl_vs_naive.png'
else:
    mu_perturb = encode_data(vib_choice, data_knn_z, exp_result.signal_names)
    labels = ['Exp20 B50',f'Exp20 knn {PERTURB_KWARGS['perturbing_sigs'][0]} {PERTURB_KWARGS['perturb_mode']}']
    filename = '/LS_fate_'  + str(len(exp_result.signal_names))  + f'D_{PERTURB_KWARGS['perturbing_sigs'][0]}_{PERTURB_KWARGS['perturb_mode']}_ctrl_vs_knn.png'

L1_perturb = mu_perturb[:, 0]
L2_perturb = mu_perturb[:, 1]

# Standardize target
L1_scaled_perturb = (L1_perturb - L1.mean())/L1.std()
L2_scaled_perturb = (L2_perturb - L2.mean())/L1.std()

rgb_image = np.zeros((200, 200, 3))

for i, fate in enumerate(fates):
    color = IIF.colmap_fates[fate]
    rgb = mcolors.to_rgb(color) # Convert color to RGB tuple (values between 0-1)
    rgb_image[i // 200, i % 200] = rgb

fig, ax = plt.subplots(n, m, figsize=(5*m, 5*n))
ax.imshow(rgb_image, origin='lower', extent=[L1smin, L1smax, L2smin, L2smax])

# Calculate KDE for the scatter points
if len(L1_scaled) > 1:  # Need at least 2 points for KDE
    # Create KDE
    values = np.vstack([L1_scaled, L2_scaled])
    kernel = sp.stats.gaussian_kde(values)
    
    # Evaluate KDE at the actual data points
    kde_at_data = kernel(values)
    
    # Set vmin so that X% of data points are outside the contours
    data_percentile = 1  # Change this: 10 means 10% of data outside, 90% inside
    vmin = np.percentile(kde_at_data, data_percentile)
    
    # Evaluate KDE on a grid
    kde_grid = kernel(np.vstack([L1grid_scaled.ravel(), L2grid_scaled.ravel()]))
    kde_grid = kde_grid.reshape(L1grid_scaled.shape)
    
    # Define levels from vmin to max
    vmax = kde_grid.max()
    levels = np.linspace(vmin, vmax, 10)
    
    # Verify
    fraction_inside = np.sum(kde_at_data >= vmin) / len(L1_scaled)
    print(f"{fraction_inside*100:.1f}% of data points are inside the contours")
    
    ax.contourf(L1grid_scaled, L2grid_scaled, kde_grid, 
                levels=levels, cmap='Greys', alpha=0.2)
    ax.contour(L1grid_scaled, L2grid_scaled, kde_grid, 
               levels=levels, colors='white', linewidths=1.5)

if len(L1_scaled_perturb) > 1:  # Need at least 2 points for KDE
    # Create KDE
    values_perturb = np.vstack([L1_scaled_perturb, L2_scaled_perturb])
    kernel_perturb = sp.stats.gaussian_kde(values_perturb)
    
    # Evaluate KDE at the actual data points
    kde_at_data_perturb = kernel(values_perturb)
    
    # Set vmin so that X% of data points are outside the contours
    data_percentile = 1  # Change this: 10 means 10% of data outside, 90% inside
    vmin_perturb = np.percentile(kde_at_data_perturb, data_percentile)
    
    # Evaluate KDE on a grid
    kde_grid_perturb = kernel_perturb(np.vstack([L1grid_scaled.ravel(), L2grid_scaled.ravel()]))
    kde_grid_perturb = kde_grid_perturb.reshape(L1grid_scaled.shape)
    
    # Define levels from vmin to max
    vmax_perturb  = kde_grid_perturb.max()
    levels_perturb = np.linspace(vmin_perturb, vmax_perturb, 10)
    
    # Verify
    fraction_inside = np.sum(kde_at_data_perturb >= vmin_perturb) / len(L1_scaled_perturb)
    print(f"{fraction_inside*100:.1f}% of data points are inside the contours")
    
    # ax.contourf(L1grid_scaled, L2grid_scaled, kde_grid_perturb, 
    #             levels=levels, cmap='Greys', alpha=0.2)
    ax.contour(L1grid_scaled, L2grid_scaled, kde_grid_perturb, 
               levels=levels_perturb, colors='black', linewidths=1.5)
    
    
ax.set_xlabel('Latent Coordinate 1',fontsize=fs)
ax.set_ylabel('Latent Coordinate 2',fontsize=fs)
ax.set_xticks([]);
ax.set_yticks([]);

for spine in ax.spines.values():
    spine.set_linewidth(sw)

ax.set_aspect('equal', adjustable='box')
ax.legend(handles=[
    Patch(facecolor=colors[0], label=labels[0]),
    Patch(facecolor=colors[1], label=labels[1])
], fontsize=12)
plt.tight_layout(pad=0.5)

plt.savefig(exp_result.data_dir + filename, dpi=150, bbox_inches='tight', pad_inches=0.1)
```
