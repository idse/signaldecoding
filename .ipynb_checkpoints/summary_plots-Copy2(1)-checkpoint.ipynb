### Result One: Single Signal Predictions

```python
# Define dataset and subset of conditions to be analyzed
mode_expt = 1
if mode_expt == 2:
    directory = 'data_expt_X10_scaled_bgsub'
    cond = 'B50'
    exp_name = 'ExpX10'
    N_sys = 4

elif mode_expt == 1:
    directory = 'data_expt_X7_scaled_bgsub'
    cond = 'B50'
    exp_name = 'ExpX7'
    N_sys = 4
  
elif mode_expt == 0:
    directory = 'data_expt_20_scaled_norm_bgsub'
    cond = 'B50'
    exp_name = 'Exp20'
    N_sys = 6

mode_reg = 'MLP'
kernel = 'relu'
```

```python
# Define a function that makes a scatter plot of the control and predicted gene expression profiles
# for this subset of genes: ISL1, TFAP2C, SOX17, TBXT, TBX6, NANOG, SOX2

# For all of these plots, we use the first colony: since we trained on the second half of colonies for all
# our predictions

subdirectory_data = directory + '/data'

# Load control data of experiment to define gene indices
f = open(subdirectory_data + "/data_B50.pickle",'rb')
data = pickle.load(f)
f.close()

gene_names = data.gene_names
gene_list = ['ISL1', 'TFAP2C', 'SOX17', 'TBXT', 'NANOG', 'SOX2']
indices_kg = [gene_names.index(g) for g in gene_list]

train_cond = 'B50'
test_cond = 'truli' #meki

# Define subdirectory to save plots to + take data from 
subdirectory_plot = directory + '/analysis_regression_sg_multi_z1_3x10/analysis_crosscond_train_'+train_cond+'_pred_' + test_cond
subdirectory_plot_data = subdirectory_plot + '/data'
dot_size = 2.5
label_fontsize = 40
tick_fontsize = 35

def scatter_plot_pred_vs_exp(exp_name, subdirectory_data, subdirectory_plot_data, train_cond, test_cond, mode_reg = 'MLP', kernel = 'relu'):

    # Load control data of experiment, to define max and min values of colormap
    f = open(subdirectory_data + "/data_B50.pickle",'rb')
    data = pickle.load(f)
    f.close()
    
    # Load colony data, including predicted and actual gene values

    
    col_num = 0
    f = open(subdirectory_plot_data + '/data_colony_it' + str(col_num) + '_' + mode_reg + kernel + "_" + test_cond + ".pickle",'rb')
    (xx_colony, yy_colony, rr_colony, markers_colony, feature_colony, target_colony, target_predict_colony) = pickle.load(f)
    f.close()


    # Predefine max and min_values
    max_vals = np.zeros(7)
    min_vals = np.zeros(7)

    # Exp 20 has a complete set of genes, while Exp X7/X10 have a reduced set
    # Here, we generalize the approach to both datasets by defining the array of 
    # indices for our list of genes
    gene_names = data.gene_names
    gene_list = ['ISL1', 'TFAP2C', 'SOX17', 'TBXT', 'NANOG', 'SOX2']
    indices_kg = [gene_names.index(g) for g in gene_list]

    fig = plt.figure(figsize=(30, 8))
    outer = gridspec.GridSpec(2, 3, wspace=0.1, hspace=0.1)
    
    for it, it_kg in enumerate(indices_kg):
        inner = gridspec.GridSpecFromSubplotSpec(1, 2,
                        subplot_spec=outer[it], wspace=0.01, hspace=0.1)

        max_vals[it] = np.nanpercentile(data.genes[:, :, it_kg], 99)
        min_vals[it] = np.nanpercentile(data.genes[:, :, it_kg], 1)

        target_colony_kg = target_colony[:, it_kg]
        target_pred_colony_kg = target_predict_colony[:, it_kg]

        # order by actual colony expression
        ind_sorted = np.argsort(target_colony_kg)
        target_colony_sorted = target_colony_kg[ind_sorted]
        target_pred_sorted = target_pred_colony_kg[ind_sorted]
        xx_colony_sorted = xx_colony[ind_sorted]
        yy_colony_sorted = yy_colony[ind_sorted]

        all_target_sorted = [target_colony_sorted, target_pred_sorted]
        
        for j in range(2):
            ax = plt.Subplot(fig, inner[j])
            if j == 0:
                ax.set_title("Exp " + gene_list[it] + ' B50', fontsize = 25)
            else:
                ax.set_title("Pred Perturb "+ gene_list[it], fontsize = 25)
                
            ax.spines['right'].set_linewidth(0)
            ax.spines['top'].set_linewidth(0)
            ax.spines['left'].set_linewidth(0)
            ax.spines['bottom'].set_linewidth(0)

        
            ax.scatter(xx_colony_sorted ,yy_colony_sorted ,dot_size,c=all_target_sorted[j] ,vmin=min_vals[it],vmax=max_vals[it],cmap='YlGnBu')
            ax.set_aspect('equal', 'box')
            ax.set_xticks([])
            ax.set_yticks([])
            fig.add_subplot(ax)    
```

```python
scatter_plot_pred_vs_exp(exp_name, subdirectory_data, subdirectory_plot_data, train_cond, test_cond)
```

```python
# Define function to compute radial profiles
def calcRadialProfiles(rr_colony, colony, indices_kg, gene_list, cellsPerBin=100, overlap=50, dr=10):
    # overlap is not really necessary if we do a Gaussian smoothing afterwards, which works better
    # dr : spacing of regular radial grid on which interpolation is performed
    
    # first create bins with equal numbers of cells and calculate the mean edgeDist & mean,std intensity in those bins
    Ncells = len(colony)
    Nbins = round(Ncells/cellsPerBin);
    edges = np.linspace(0, Ncells-1, Nbins+1).round().astype(int);

    
    I = np.argsort(rr_colony)
    r_tmp = np.zeros(Nbins)
    profile_tmp = pd.DataFrame(columns= gene_list , index=range(Nbins))
    profile_tmp_std = pd.DataFrame(columns= gene_list , index=range(Nbins))
    
    for i in range(Nbins):
        
        start = max(edges[0], edges[i]-overlap)
        stop = min(edges[i+1]+overlap, edges[-1])
        ptidx = np.arange(start, stop+1) 
        
        sel = I[ptidx]
        r_tmp[i] = np.mean(rr_colony[sel])
        profile_tmp.loc[i, :] = np.nanmean(colony[:, indices_kg][sel, :], axis=0)
        profile_tmp_std.loc[i, :] = np.nanstd(colony[:, indices_kg][sel, :], axis=0)

    r_tmp[-1] = np.max(rr_colony); # for center bin, the r value should be the center, not the average r value of the points in it

    
    # now interpolate on evenly spaced radial bins that allow easy averaging between colonies
    radius_micron = 350
    margin = radius_micron/10;
    maxR = radius_micron + margin
    # I am adding 10% negative R values to the grid on which it will linearly extrapolate, to deal with boundary effects for smoothing on the edge 
    # (otherwise positional error goes high on edge because mirroring the data for smoothing makes the gradient zero there)
    Ngrid = round((maxR + margin)/dr)
    radial_Grid = np.linspace(-margin, maxR, Ngrid)
    
    radial_Profiles = pd.DataFrame(columns=gene_list, index=range(Ngrid))
    radial_Profiles_std = pd.DataFrame(columns=gene_list, index=range(Ngrid))

    
    for gene in gene_list:
        
        interpolator = interp.interp1d(r_tmp, profile_tmp[gene], kind='linear', fill_value='extrapolate', bounds_error=False)
        radial_Profiles[gene] = interpolator(radial_Grid);

        interpolator = interp.interp1d(r_tmp, profile_tmp_std[gene], kind='linear', fill_value='extrapolate', bounds_error=False)
        radial_Profiles_std[gene] = interpolator(radial_Grid);

    return radial_Grid, radial_Profiles
```

```python
# Compute radial profiles of colonies: ctrl vs perturbation
all_exp_profiles = []
all_pred_profiles = []
all_perturb_profiles = []
    

for it in range(N_sys):
    f = open(subdirectory_plot_data + '/data_colony_it' + str(it) + '_' + mode_reg + kernel + "_" + test_cond + ".pickle",'rb')
    (xx_colony, yy_colony, rr_colony, markers_colony, feature_colony, target_colony, target_predict_colony) = pickle.load(f)
    f.close()

    radial_Grid, radial_Profiles_exp = calcRadialProfiles(rr_colony, target_colony, indices_kg, gene_list)
    _, radial_Profiles_pred = calcRadialProfiles(rr_colony, target_predict_colony, indices_kg, gene_list)

    all_exp_profiles.append(radial_Profiles_exp)
    all_pred_profiles.append(radial_Profiles_pred)

    # Here, we only plot the control + perturbation condition if the perturbation tested matches with the experiment
    if exp_name == 'ExpX7' and test_cond == 'truli':
        subdirectory_perturb = directory + '/analysis_regression_sg_multi_z1_3x10/analysis_crosscond_train_B50T30_pred_B50T30'
        subdirectory_perturb_data = subdirectory_perturb + '/data'
        
        f = open(subdirectory_perturb_data + '/data_colony_it' + str(it) + '_' + mode_reg + kernel + "_B50T30.pickle",'rb')
        (xx_colony, yy_colony, rr_colony, markers_colony, feature_colony, target_colony, target_predict_colony) = pickle.load(f)
        f.close()

        _, radial_Profiles_perturb = calcRadialProfiles(rr_colony, target_colony, indices_kg, gene_list)
        all_perturb_profiles.append(radial_Profiles_perturb)
        
    elif exp_name == 'ExpX10' and test_cond == 'meki':
        subdirectory_perturb = directory + '/analysis_regression_sg_multi_z1_3x10/analysis_crosscond_train_B50_MEKi_pred_B50_MEKi'
        subdirectory_perturb_data = subdirectory_perturb + '/data'
        
        f = open(subdirectory_perturb_data + '/data_colony_it' + str(it) + '_' + mode_reg + kernel + "_B50_MEKi.pickle",'rb')
        (xx_colony, yy_colony, rr_colony, markers_colony, feature_colony, target_colony, target_predict_colony) = pickle.load(f)
        f.close()
    
        _, radial_Profiles_perturb = calcRadialProfiles(rr_colony, target_colony, indices_kg, gene_list)
        
        all_perturb_profiles.append(radial_Profiles_perturb)
        
```

```python
mean_exp_profiles = pd.concat(all_exp_profiles).groupby(level=0).mean()
std_exp_profiles = pd.concat(all_exp_profiles).groupby(level=0).std()

mean_pred_profiles = pd.concat(all_pred_profiles).groupby(level=0).mean()
std_pred_profiles = pd.concat(all_pred_profiles).groupby(level=0).std()

# If plotting control + perturbations
if len(all_perturb_profiles)>0:
    mean_perturb_profiles = pd.concat(all_perturb_profiles).groupby(level=0).mean()
    std_perturb_profiles = pd.concat(all_perturb_profiles).groupby(level=0).std()
```

```python
plt.figure(figsize = (30, 13))
for it_gene, gene in enumerate(gene_list):  
    plt.subplot(2, 3, it_gene + 1)
    plt.title(gene_list[it_gene], fontsize = 30)
    ax = plt.gca()


    plt.plot(radial_Grid, mean_pred_profiles[gene],'-',color= 'red', label= test_cond + ' pred', lw=6)
    plt.fill_between(radial_Grid, mean_pred_profiles[gene] - std_pred_profiles[gene] , mean_pred_profiles[gene] + std_pred_profiles[gene], color= 'red', alpha=0.2)
    if len(all_perturb_profiles)>0:
        plt.plot(radial_Grid, mean_perturb_profiles[gene],'-',color= 'blue', label= test_cond + ' exp', lw=6)
        plt.fill_between(radial_Grid, mean_perturb_profiles[gene] - std_perturb_profiles[gene] , mean_perturb_profiles[gene] + std_perturb_profiles[gene], color= 'blue', alpha=0.2)
        plt.plot(radial_Grid, mean_exp_profiles[gene],'-',color= 'gray', label= 'B50 exp', lw=6)
        plt.fill_between(radial_Grid, mean_exp_profiles[gene] -std_exp_profiles[gene] , mean_exp_profiles[gene] +std_exp_profiles[gene] ,color= 'gray', alpha=0.2)
    else:
        plt.plot(radial_Grid, mean_exp_profiles[gene],'-',color= 'blue', label= 'B50 exp', lw=6)
        plt.fill_between(radial_Grid, mean_exp_profiles[gene] -std_exp_profiles[gene] , mean_exp_profiles[gene] +std_exp_profiles[gene] ,color= 'blue', alpha=0.2)
        
    ax.set_xticks(ticks = np.arange(0, 351, step=70), labels = [0, '', '','', '', 350], fontweight = 'normal', fontsize=tick_fontsize ,  family='Arial' , color = 'black')  
    ax.set_box_aspect(1)

    plt.xlabel('edge dist. (μm)', fontsize = label_fontsize, fontweight = 'normal', family='Arial')
    
    plt.gca().tick_params(axis='both', which='major', labelsize=tick_fontsize, width=3, length=10)
    
    plt.gca().spines['right'].set_linewidth(0)
    plt.gca().spines['top'].set_linewidth(0)
    legend = ax.legend( prop={'size': 25, 'family': 'Arial', 'weight': 'normal'}, loc = 'upper right')
    ax.set(xlim = (0,350), ylim = (0, None))
plt.tight_layout()
```
